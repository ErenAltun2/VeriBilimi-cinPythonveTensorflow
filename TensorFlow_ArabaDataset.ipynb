```python
dataFrame=dataFrame.drop("transmission",axis=1)
```

```python
dataFrame
```

```python
y=dataFrame["price"].values
x=dataFrame.drop("price",axis=1).values
```

```python
y
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=10)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
X_test=scaler.transform(X_test)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
X_train.shape
```

```python
model=Sequential()
model.add(Dense(12,activation="relu"))
model.add(Dense(12,activation="relu"))
model.add(Dense(12,activation="relu"))
model.add(Dense(12,activation="relu"))
model.add(Dense(1))
model.compile(optimizer="adam",loss="mse") #buraya metrics yazıp accuracy yazsaydım eger kı accuracy nı de gorecektım
```

```python
model.fit(X_train,y_train,validation_data=(X_test,y_test),epochs=300,batch_size=250)
```

```python
kayipVerisi=pd.DataFrame(model.history.history)
```

```python
kayipVerisi.head()
```

```python
kayipVerisi.plot()
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
tahminDizisi=model.predict(X_test)
```

```python
tahminDizisi
```

```python
mean_absolute_error(y_test,tahminDizisi)  #mae ortalama mutlak hata hesabı yapıyoruz yanı gercek sonuclar ıle tahmın sonuclarını cıkarıyoruz durmadan ve ortalamasını alıyoruz.
```
