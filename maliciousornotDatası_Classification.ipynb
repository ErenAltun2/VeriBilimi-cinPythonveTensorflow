```python
y=dataFrame["Type"].values
x=dataFrame.drop("Type",axis=1).values
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=15)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train=scaler.transform(X_train)
X_test=scaler.transform(X_test)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Activation,Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
X_train.shape
```

```python
model = Sequential()
model.add(Dense(units=30,activation="relu"))
model.add(Dense(units=15,activation="relu"))
model.add(Dense(units=15,activation="relu"))
model.add(Dense(units=1,activation="sigmoid"))

model.compile(loss="binary_crossentropy",optimizer="adam") #loss kayıp fonksıyonunu sınıflandırma oldugu ıcın bınary crossentropy sectık normalde regresyonlarda mae ve mse secebılırırz
```

```python
model.fit(x=X_train,y=y_train,epochs=600,validation_data=(X_test,y_test),verbose=1)
```

```python
model.history.history
```

```python
modelKaybi=pd.DataFrame(model.history.history)
```

```python
modelKaybi.plot()
```

```python
#overfitting olmus burda egıtılme verısınde loss degerı dusuyor yanı model ogrenıyor gıbı gorunuyor ama fazla ogrenmıs tahmın verılerınde valudatıon loss artmıs sapma sapan bır yerde
```

```python
model = Sequential()
model.add(Dense(units=30,activation="relu"))
model.add(Dense(units=15,activation="relu"))
model.add(Dense(units=15,activation="relu"))
model.add(Dense(units=1,activation="sigmoid"))

model.compile(loss="binary_crossentropy",optimizer="adam") #loss kayıp fonksıyonunu sınıflandırma oldugu ıcın bınary crossentropy sectık normalde regresyonlarda mae ve mse secebılırırz
```

```python
earlyStopping=EarlyStopping(monitor="val_loss",mode="min",verbose=1,patience=25)
```

```python
model.fit(x=X_train,y=y_train,epochs=700,validation_data=(X_test,y_test),verbose=1,callbacks=[earlyStopping])
```

```python
modelKaybi=pd.DataFrame(model.history.history)
```

```python
modelKaybi.plot()
```

```python
#callback cagrısıyla bırlıkte orada val loss a monıtorıng dıyerek onu ınceledı kotulesmeye basladıgını anladıgı an durdurdu modelı egıtmeyı
```

Dropout

```python
model = Sequential()
model.add(Dense(units=30,activation="relu"))
model.add(Dropout(0.5))
model.add(Dense(units=15,activation="relu"))
model.add(Dropout(0.5))
model.add(Dense(units=15,activation="relu"))
model.add(Dropout(0.5))
model.add(Dense(units=1,activation="sigmoid"))

model.compile(loss="binary_crossentropy",optimizer="adam")
```

```python
model.fit(x=X_train,y=y_train,epochs=700,validation_data=(X_test,y_test),verbose=1,callbacks=[earlyStopping])
```

```python
kayipDf=pd.DataFrame(model.history.history)
```

```python
kayipDf.plot()
```

```python
tahminlerimiz=model.predict(X_test)
```

```python
tahminlerimiz=(tahminlerimiz>0.5).astype("int32")
```

```python
tahminlerimiz
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test,tahminlerimiz))
```

```python
print(confusion_matrix(y_test,tahminlerimiz))
```
