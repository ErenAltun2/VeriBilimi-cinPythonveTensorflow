#Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
yasListesi=[10,20,30,40,50,60,70,75]
kiloListesi=[20,60,80,85,86,87,70,90]
```

```python
numpyYasListesi=np.array(yasListesi)
numpyKiloListesi=np.array(kiloListesi)
```

```python
numpyKiloListesi
```

```python
numpyYasListesi
```

```python
plt.plot(numpyYasListesi,numpyKiloListesi,"g") #ılk verdıgımız deger x eksenıne yazılır plt.plot ıle cızımı yaptırırız buradakı g ıse renk green cızgı ıle goster demek
plt.show()  #plt.show ıle cızımı gosterırız
```

```python
plt.plot(numpyKiloListesi,numpyYasListesi,"r") #x ile y nın yerını degıstırdık gırdı olarak kıloyu basta verdık ama kafa karısıyor
plt.show()
```

```python
plt.plot(numpyYasListesi,numpyKiloListesi,"r")
plt.xlabel("Yas")  #x ve y eksenıne etiket ısımlerı verebılıyoruz xlabel ve ylabel dıyerek
plt.ylabel("Kilo")
plt.title("kilonun yasa gore degisimi grafiği") #grafıge baslık koyabılırız
plt.show()
```

```python
numpyDizisi1=np.linspace(0,10,20)
numpyDizisi1
```

```python
numpyDizisi2=numpyDizisi1**2
```

```python
plt.plot(numpyDizisi1,numpyDizisi2,"g--") #g nın yanına koydugum -- cızgı ıle kesık kesık gosterır *- koyarsam ıle yıldızlı gosterır verılerın oldugu yerı
plt.show()
```

```python
plt.plot(numpyDizisi1,numpyDizisi2,"g-*")
plt.show()
```

```python
plt.subplot(1,2,1) #alt grafık cızdırmeye yarıyor  ılk gırılen deger kac satır olacak 2. verılen deger kac sutun olacak kacıncı grafık 1 dıye verıyoruz
plt.plot(numpyDizisi1,numpyDizisi2,"g*-")
plt.subplot(1,2,2) #1 sıra olacak 2 kolon olacak şuan 2.grafıgı cızıyorum anlamına gelıyor
plt.plot(numpyDizisi1,numpyDizisi2,"r--")
plt.show()
#
```

```python
benimFigure=plt.figure() #bos bır figure oluyor yanı bos bır sayfa tuval gıbı dusun plt.plot veya plt.subplot hazır kutuphane metotlarını kullanarak bızden ekstra degerler almadan otomatık olarak cızdırıyor figure ıle daha ozellestırerek cızdırebılıyoruz
figureAxes=benimFigure.add_axes([0.2,0.2,0.9,0.9])
figureAxes.plot(numpyDizisi1,numpyDizisi2,"g")
figureAxes.set_xlabel("X Ekseni")
figureAxes.set_ylabel("Y Ekseni")
figureAxes.set_title("Baslık")
plt.show()
```

```python
figure2=plt.figure() #matplotlibden figure nesnesı uretıyoruz
eksen1=figure2.add_axes([0.1,0.1,0.7,0.7])
eksen2=figure2.add_axes([0.25,0.4,0.1,0.2])  #buradakı ılk ıkı deger yanı 0.2 ve 0.4 sırasıyla bos kagıttakı yanı olusan fıgure da koordınatları gırıyoruz en sol kısım orıjın oluyor 0,0 noktası sonrasında x,y koordınatlarını gırıyoruz
eksen1.plot(numpyDizisi1,numpyDizisi2,"g")  #0.1,0.2 son ıkılı kısım ıse genıslık ve uzunluk seklınde oluyor mesela en son 0.2 ye 0.4 yaptım alt tarafı kısa yukseklıgı uzun oldu U gıbı oldu
                                            # yanı ılk durum x,y koordinatını dıger ıkılı durum ıse xy koselerı bellı edıyor bıze
eksen1.set_xlabel("X Ekseni")
eksen1.set_ylabel("Y Ekseni")
eksen1.set_title("Baslık")

eksen2.plot(numpyDizisi2,numpyDizisi1,"b")
eksen2.set_xlabel("X Ekseni")
eksen2.set_ylabel("Y Ekseni")
eksen2.set_title("KucukGrafikBaslık")
plt.show()
```

1. plt.subplot() (Tekil - Eski / MATLAB Tarzı)
Her bir alt grafik için bu fonksiyonu ayrı ayrı çağırman gerekir. Argüman olarak (satır, sütun, kaçıncı_grafik) alır ve o an seçilen grafiği aktif hale getirir.

2. plt.subplots() (Çoğul - Modern / Nesne Yönelimli Tarz)
Hem tuvali (fig) hem de tüm eksenleri (axes) tek seferde üretir. Eksenler bir NumPy dizisi olarak döner; böylece axes[0], axes[1] diyerek doğrudan istediğin grafiğe müdahale edersin.

```python
(benimFigure,benimEksen)=plt.subplots(nrows=1,ncols=2)
type(benimEksen)
for eksen in benimEksen:
    eksen.plot(numpyDizisi1,numpyDizisi2,"g")
    eksen.set_xlabel("X Ekseni")
    eksen.set_ylabel("Y Ekseni")
    eksen.set_title("Baslık")
plt.tight_layout() #grafıkler arası mesafeyı duzeltır.
plt.show()
```

ıkı farklı tuvale cızmek ıcın ıkı kere fıgure yazarım
aynı tuval ıcıne cızmek ıcın 3 tane yontem var
1-figure kullanıp axes ıle koordınatlarını degıstırerek yapabılırız
2-plt.subplot ıle matplotlib in eskı yontemı ıle cızdırebılırız dırekt 1,2,1 dıyerek 1 satır 2 sutun olsun 1.grafık seklınde kullanırız
3- plt.subplots dıyerek nrows ne ncols seklınde kac satır kac sutun oldugunu belırlerız grafıklere kacıncı sutun kacıncı satır seklınde erısım saglarız

fig, axes = plt.subplots(nrows=2, ncols=2)

# 1. Satır, 1. Sütun (Sol Üst)
axes[0, 0].plot(x, y)

# 1. Satır, 2. Sütun (Sağ Üst)
axes[0, 1].plot(x, y)

# 2. Satır, 1. Sütun (Sol Alt) -> Aradığın yer tam olarak burası
axes[1, 0].plot(x, y)

# 2. Satır, 2. Sütun (Sağ Alt)
axes[1, 1].plot(x, y)

bu sekılde ulasırız

#Matplotlib Görsellik

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
numpyDizisi1=np.linspace(0,10,20)
```

```python
numpyDizisi1
```

```python
numpyDizisi2=numpyDizisi1**2
```

```python
(benimFigur,benimEksen)=plt.subplots()  #benımfıgur dedıgımız sey aslında genel gorunen gorsel beyaz tuvel oluyor zaman benım eksen dedııgımız ıse cızdırdıgımız grafıkler oluyor
benimEksen.plot(numpyDizisi1,numpyDizisi2,"g",alpha=0.3) #alpha saydamlık katar
benimEksen.plot(numpyDizisi2,numpyDizisi1,"r")
plt.show()
```

```python
(yeniFigur,yeniEksen)=plt.subplots()
yeniEksen.plot(numpyDizisi1,numpyDizisi2+2,color="blue",alpha=0.3,linewidth=5.0)
yeniEksen.plot(numpyDizisi2,numpyDizisi1+2,"r")
yeniEksen.plot(numpyDizisi1,numpyDizisi2+10,color="#890000",linestyle="-")
yeniEksen.plot(numpyDizisi2,numpyDizisi1+20,"b",linestyle="--")
yeniEksen.plot(numpyDizisi2,numpyDizisi1+30,color="#000000",linestyle="--",marker="o",markersize=4,markerfacecolor="red")
yeniEksen.plot(numpyDizisi2,numpyDizisi1+40,color="#000000",linestyle="--",marker="*",markersize=4,markerfacecolor="red")
plt.show()

"""
Parantezin içini boş bıraktığında plt.subplots() varsayılan olarak nrows=1, ncols=1 üretir.
Yani elinde sadece 1 satır ve 1 sütundan oluşan tek bir grafik alanı olur; yanına ya da altına başka bir kutu/grafik ekleyemezsin.
"""
```
