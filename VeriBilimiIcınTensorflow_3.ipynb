##instance & attribute

Instance (Örnek): Tespitin kesinlikle doğru. benim_listem = list() dediğinde, soyut list sınıfından bellekte somut bir örnek (instance/object) üretmiş olursun.

Attribute (Nitelik / Özellik): Nesnenin taşıdığı verilerdir (durum/state). Örneğin bir araba nesnesi için renk = "kırmızı" veya hiz = 100 gibi değerler birer attribute'tur.

Method (Metot): Nesnenin yapabildiği davranışlardır (fonksiyonlar). Yani "eriştiğimiz özellikler" değil, nesneye bağlı olarak çalışan eylemlerdir.

Fonksiyon: Bağımsızdır, herhangi bir sınıfa veya nesneye doğrudan ait olmak zorunda değildir. Doğrudan çağrılır: len(benim_listem) ya da print().

Metot: Bir sınıfa/nesneye aittir. Nesne üzerinden çağrılır ve genellikle o nesnenin verisini (self) işler: benim_listem.append(5).

```python
class SuperKahraman(): #sınıfların genelde ılk harflerı buyuk yazılır
  ozelGuc="Gorunmezlik"
  def __init__(self,isimInput,yas,meslekInput): # initialization dan gelmekte baslangıc fonksıyonu yanı constructor
   print("init çağrıldı")
   self.isim=isimInput
   self.yas=yas
   self.meslekInput=meslekInput
  def ornekMethod(self):
    print(f"ben süper kahramanım ve mesleğim {self.meslekInput}")
```

```python
class Kopek():
  yilCarpani=7
  def __init__(self,yas=5):  #default degerı aynı fonksıyonda yaptıgımız gıbı yapıyoruz.
    self.yas=yas

  def insanYasi(self):
    return self.yas*Kopek.yilCarpani #veya self.yilCarpani seklınde kullanabılırız dırekt yilCarpani seklınde kullanamayız ama cunku o ankı olusturdugumuz nesnenın oldugunu belırtmemız gerekıyor

  """
Fonksiyonlar Global kapsama doğrudan bakabilir ama Class (Sınıf) kapsamına doğrudan bakamaz.

Dışarıda tek başına duran isim = "eren" bir Global değişkendir. Fonksiyon onu doğrudan okuyabilir.

Ancak class Kopek: içine yazdığın yilCarpani = 7 bir global değişken değildir; o sınıfın sınırları içine hapsedilmiş bir Sınıf Değişkenidir.

Bu yüzden Python metot içindeyken sınıf gövdesine dönüp bakmaz; ona ulaşmak için adresi açıkça belirtmeni (Kopek.yilCarpani veya self.yilCarpani) ister.
  """
```

```python
benimKopek=Kopek(3)
```

```python
benimKopek.yas
```

```python
benimKopek.insanYasi()
```

#inheritance

```python
class Hayvan():
  def __init__(self):
    print("hayvan sınıfı init çağrıldı")
  def method1(self):
    print("hayvan sınıfı method 1 cagırıldı")
  def method2(self):
    print("hayvan sınıfı method 2 cagırıldı")
```

```python
benimHayvanim=Hayvan()
```

```python
benimHayvanim.method1()
```

```python
benimHayvanim.method2()
```

```python
class Kedi(Hayvan):
  def __init__(self):
    Hayvan.__init__(self)
    print("kedi sınıfı init çağrıldı")
  def miyavla(self):
    print("miyav")
  #Override
  def method1(self):
    print("kedi sınıfındaki method1 çağrıldı")
```

```python
benimKedi=Kedi()
```

```python
benimKedi.method1() #eskısı bu method1 i kedi sınıfında yapmadan once
```

```python
benimKedi.miyavla()
```

```python
benimKedi.method1() #method1 i kedi sınıfı ıcın duzenledıkten sonra
```

```python
class Elma():
  def __init__(self,isim):
    self.isim=isim
  def bilgiVer(self):
    return self.isim + " 100 kaloridir"
```

```python
class Muz():
  def __init__(self,isim):
    self.isim=isim
  def bilgiVer(self):
    return self.isim+ " 150 kaloridir"
```

```python
elma = Elma("elma")
```

```python
elma.bilgiVer()
```

```python
muz=Muz("muz")
```

```python
muz.bilgiVer()
```

```python
meyveListesi=[elma,muz] #meyve lıstesının ıcıne objelerı koyuyorum.
```

```python
for meyve in meyveListesi:
  print(meyve.bilgiVer())
```

```python
def bilgiAl(meyve):
  print(meyve.bilgiVer())
```

#**OzelMetodlar**

```python
class Meyve():
  def __init__(self,isim,kalori):
    self.isim=isim
    self.kalori=kalori
  def __str__(self):
    return f"{self.isim} şu kadar kaloriye sahiptir: {self.kalori}"
  def __len__(self):
    return self.kalori
```
