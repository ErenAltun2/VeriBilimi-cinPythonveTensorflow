pandasla ayrı olan dataları bırlestırme ıslemı
1- aynı sutunlara sahıp ama farklı ıkı verıyı yanı sutun baslıkları aynı ıkısınde de ısımler ve sporlar olsun onu bırlestırmeyı gordum

```python
sozluk1={"Isim":["Ahmet","Mehmet","Zeynep","Atil"],
         "Spor":["Koşu","Yüzme","Koşu","Basketbol"],
         "Kalori":[100,200,300,400]
         }
```

```python
dataFrame1=pd.DataFrame(sozluk1,index=[0,1,2,3])
```

```python
dataFrame1
```

```python
sozluk2={"Isim":["Osman","Levent","Atlas","Fatma"],
         "Spor":["Koşu","Yüzme","Badmintoon","Tenis"],
         "Kalori":[200,100,50,300]
         }
```

```python
dataFrame2=pd.DataFrame(sozluk2,index=[4,5,6,7])
```

```python
dataFrame2
```

```python
sozluk3={"Isim":["Ayse","Mahmut","Duygu","Nur"],
         "Spor":["Koşu","Yüzme","Badmintoon","Tenis"],
         "Kalori":[300,600,500,200]
         }
```

```python
dataFrame3=pd.DataFrame(sozluk3,index=[8,9,10,11])
```

```python
dataFrame3
```

```python
#concatenation  alt alt bırlestırme kolonlar aynı ise kullanılıyorsa
```

```python
pd.concat([dataFrame1,dataFrame2,dataFrame3],axis=0) #pd.concat([dataFrame1,dataFrame2,dataFrame3],axis=1) dersek egerkı yanyana bırlestırır axis yazmassak otomatık olarak axis=0 alır
```

##merge iki farklı datayı yan yana bırlestırır ama bır sutunu secerız ve onu anahtar belırlerız bu sayede anahtar belırledıgımız degere gore esleme yapar ve bırlestırır

```python
mergeDataFrame1=pd.DataFrame(mergeSozluk1)
```
